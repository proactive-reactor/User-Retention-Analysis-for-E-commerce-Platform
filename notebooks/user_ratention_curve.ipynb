```python
#convert date
df['signup_date'] = pd.to_datetime(df['signup_date'])
df['activity_date'] = pd.to_datetime(df['activity_date'])
```

```python
#calculate cohort index
df['cohort_index'] = (df['activity_date'] - df['signup_date']).dt.days
```

```python
#pivot table
cohort = df.pivot_table(
    index='signup_date',
    columns='cohort_index',
    values='retained_users'
)
```

```python
cohort.iloc[:, :10].plot(figsize=(10,6))
plt.title("User Retention Curve")
plt.show()
```
