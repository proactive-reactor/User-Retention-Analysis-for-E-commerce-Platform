```python
fig, ax1 = plt.subplots()
# fig, ax = plt.subplots()
```

```python
# ax.bar(df['date'], df['pageviews'], color='lightgray', label='Pageview')
```

```python
# Plot bars on primary axis
ax1.bar(df['date'], df['pageviews'], color='lightgray', label='Pageview')
ax1.set_ylabel('Pageviews', color='gray')
```

```python
# ax.plot(df['date'], df['transactions'], color='blue', marker='o', label='Growth %')
```

```python
# Create a second axes for the line plot
ax2 = ax1.twinx()
ax2.plot(df['date'], df['transactions'], color='blue', marker='o', label='Growth %')
ax2.set_ylabel('Transactions/Growth %', color='blue')
```

```python
# Add legends
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')
```
